```python
import rocks 
import matplotlib.pyplot as plt
import pandas as pd 

df = pd.read_csv('/Users/chloejgordon14/Downloads/UPDATED sheet2 - Sheet2-3.csv')
df['clean_name'] = [name for name, number in rocks.id(df['name'])]

bft = rocks.load_bft();

matched_df = pd.merge(
    df,
    bft[['name', 'diameter.value',
         'a']],
    left_on='clean_name',
    right_on='name',
    how='left'
)

x = matched_df['diameter.value']
y = matched_df['a']

plt.scatter(x,y)

plt.scatter(x, y, color='xkcd:pale blue')
plt.xlabel('Diameter')
plt.ylabel('Semi-major axis (a)')
plt.title('Asteroid Diameter vs. Semi-major Axis')
```
