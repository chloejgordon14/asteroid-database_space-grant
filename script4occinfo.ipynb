# Angular Separation v.s. Physical Separation

```python
import matplotlib.pyplot as plt 
import pandas as pd
import rocks 

df = pd.read_csv('/Users/chloejgordon14/Downloads/Copy of asteroid_satellites for Occ Info - Sheet2.csv')

df.columns = df.columns.str.strip()

df = df.dropna(subset=['name']).copy()
df['clean_name'] = [name for name, number in rocks.id(df['name'])]

matched_df = pd.merge(df, bft[['name']], left_on='clean_name', right_on='name', how='left');



matched_df=matched_df.dropna(subset='name_x')

matched_df['angular sep'] = matched_df['angular sep'].astype(str).str.replace('"', '', regex=False).str.strip()
matched_df['angular sep'] = pd.to_numeric(matched_df['angular sep'], errors='coerce')
matched_df['physical sep'] = pd.to_numeric(matched_df['physical sep'], errors='coerce')


plot_mask = ( matched_df['angular sep'].notna() &   matched_df['physical sep'].notna() & (matched_df['name_x'] != "queens"))

x = matched_df.loc[plot_mask, 'angular sep']
y = matched_df.loc[plot_mask, 'physical sep']

plt.figure(figsize=(10, 6))
plt.scatter(x,y, marker='o', color='xkcd:pumpkin')
plt.xlabel('angular separation')
plt.ylabel('physical separation')

```
