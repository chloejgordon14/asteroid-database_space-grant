```python
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_csv('/Users/chloejgordon14/Downloads/disco_clean - disco_clean-2.csv')

technique_counts = df['technique'].value_counts()


x = technique_counts.index.astype(str)
y = technique_counts.values

plt.clf()
plt.figure(figsize=(12,6))
plt.bar(x,y, color='xkcd:rose pink')

plt.xlabel('Technique', fontsize=12)
plt.ylabel('Number of Asteroids Found', fontsize=12)
```
