```python
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_csv('/Users/chloejgordon14/Downloads/disco_clean - disco_clean-2.csv')

df_clean = df.dropna(subset=['technique', 'discoverer'])

top_discoverers = df_clean['discoverer'].value_counts().head(10).index

df_top = df_clean[df_clean['discoverer'].isin(top_discoverers)]

count_table = pd.crosstab(df_top['technique'], df_top['discoverer'])

plt.clf()
count_table.plot(kind='bar', stacked=True, figsize=(12, 6), colormap='tab20')

```
